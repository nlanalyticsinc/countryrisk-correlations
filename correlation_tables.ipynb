## Summary Table

The table below is rendered from the generated CSV outputs.

```python
import csv
from pathlib import Path

try:
    from IPython.display import Markdown, display
except ImportError:
    Markdown = None
    display = None

def read_csv(path):
    with Path(path).open(newline='') as handle:
        return list(csv.DictReader(handle))

variant = 'filled'
gvkey_quarter = [row for row in read_csv('data/correlations/gvkey_quarter_correlations.csv') if row['variant'] == variant]
quarter = {row['country_code']: row for row in read_csv('data/correlations/quarter_correlations.csv') if row['variant'] == variant}

lines = [
    f'| Country | Variant | Firm-quarter Pearson | Quarter Pearson | Matched firm-quarters | Quarters |',
    '|---|---|---:|---:|---:|---:|',
]
for row in gvkey_quarter:
    qrow = quarter[row['country_code']]
    lines.append(
        f"| `{row['country_code'].upper()}` | "
        f"{variant} | "
        f"{float(row['pearson']):.2f} | "
        f"{float(qrow['pearson']):.2f} | "
        f"{int(row['n_pairs']):,} | "
        f"{int(qrow['n_pairs']):,} |"
    )

table = '\n'.join(lines)
if display is not None:
    display(Markdown(table))
else:
    print(table)
```
